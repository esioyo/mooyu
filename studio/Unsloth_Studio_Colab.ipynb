### 直接下载文件到 Google Drive
使用 `wget` 工具可以将模型或数据集直接从 URL 下载到您的云端硬盘。

```python
# 1. 安装必要的库
!pip install -q huggingface_hub

from huggingface_hub import snapshot_download
import os

# 2. 设置模型 ID 和保存路径
model_id = "HauhauCS/Gemma-4-E4B-Uncensored-HauhauCS-Aggressive"
save_path = "/content/drive/MyDrive/models/Gemma-4-E4B-Uncensored"

# 3. 开始下载 (这将下载整个模型权重和配置文件)
os.makedirs(save_path, exist_ok=True)

print(f"正在下载模型 {model_id} 到 {save_path}...")
snapshot_download(
    repo_id=model_id,
    local_dir=save_path,
    local_dir_use_symlinks=False
)

print(f"下载完成！")
```
